```python
from IPython.display import display
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.backends.backend_pdf import PdfPages
from skimage import io, filters, measure, morphology, util, color
from skimage.measure import find_contours
from pathlib import Path
from scipy.spatial import ConvexHull
from scipy import ndimage as ndi
from skimage.segmentation import clear_border
import time

import skimage.transform as transform
```

```python
# --- Folder ---
folder = Path(r'C:\Users\jason.orlando\OneDrive - Sun Chemical\Documents\TEM\inprog request data\REQ2603473 characterization of 449-0519 lot CIUC0139\50kx')
```

<h1 style=" font-size:20px;">
    Image initial initial variable suggestion
</h1>

```python
# --- 1. FILE SELECTION ---
target_file_name = None  #Set to None to automatically grab the first acceptable file_type.
file_type = "*.jpg"

# --- 2. FILE RESOLUTION LOGIC ---
plt.close('all')
if target_file_name is not None:
    active_file = folder / target_file_name
else:
    active_file = next(folder.glob(file_type), None)

# --- 3. GEOMETRY SUGGESTION ENGINE ---
_img_raw = io.imread(active_file)
_img_temp = color.rgb2gray(_img_raw) if _img_raw.ndim == 3 else _img_raw
_img_temp = util.img_as_float32(_img_temp)
img_height, img_width = _img_temp.shape

_test_mask = _img_temp < filters.threshold_otsu(_img_temp)
_dist_transform = ndi.distance_transform_edt(_test_mask)
_max_thickness = np.max(_dist_transform)

_suggested_bg_blur = int(max(img_width, img_height) * 0.10)
_suggested_block = int(_max_thickness * 4)
if _suggested_block % 2 == 0: _suggested_block += 1
_suggested_block = max(21, min(_suggested_block, 151))

_suggested_offset = round(float(np.std(_img_temp) * 0.12), 3)
_est_particle_radius = max(3.0, _max_thickness * 0.5)
_suggested_min_area = int((_est_particle_radius ** 2) * np.pi)
_suggested_watershed_dist = int(_est_particle_radius * 1.5)

print("\n" + "="*60)
print(f"   📊 SUGGESTED BASELINES FOR: {active_file.name}")
print("="*60)
print(f" -> background_blur_sigma: {_suggested_bg_blur} px")
print(f" -> gaussian_sigma:         1")
print(f" -> adaptive_block_size:    {_suggested_block} px")
print(f" -> adaptive_offset:        {_suggested_offset}")
print(f" -> min_particle_area:      {_suggested_min_area} px")
print(f" -> watershed_min_distance: {_suggested_watershed_dist} px")
print("="*60 + "\n")
```

<div id="edu"></div>
<h1 style=" font-size:20px;">
    Particle or aggregate counting and measuring
</h1>

```python
# =======================# 2. TUNABLE PARAMETERS====================================================
adaptive_offset =  _suggested_offset * 1                              # if offset = 0 a particle just must be above local mean (Threshold = local mean - offset)
min_particle_area = int(_suggested_min_area * 1)
first_pass = .4
aggregate_bridging_factor = 3                                           # higher numbers bridge more gaps

adaptive_block_size = int(_suggested_block * 1)                         # pixel diameter for relative brightness [otsu would be adaptive_block_size = whole image and adaptive_offset = 0]
if adaptive_block_size % 2 == 0: adaptive_block_size += 1

# Image smoothing
gaussian_sigma = 1                                                      # Higher values smooth rough surfaces and blend noise
background_blur_sigma = _suggested_bg_blur                              # pixel diameer, Must > largest aggregate size, subtracts local background
  
# watershed particle separation [only toggle for particles]
use_watershed = True                                                    # True = slice clusters into particles
watershed_min_distance = int(_suggested_watershed_dist * 1)           # particles wont be sliced up so they their centers are closer than this distance

# Data Integrity Controls
exclude_border_particles = True                                        # True removes chopped particles touching image borders
scale_bar_zone_fraction = 0.03                                         # Fraction of the bottom screen to erase (0.05 = bottom 5%)
# ==================================================================================================
# 3. IMAGE PROCESSING & MEASUREMENT PIPELINE
# ==================================================================================================

start = time.time()
total_time_for_image = 0

image_raw = io.imread(active_file)
if image_raw.ndim == 3:
    image = color.rgb2gray(image_raw)
else:
    image = image_raw
image = util.img_as_float32(image)
img_height, img_width = image.shape
label_offset = img_height * scale_bar_zone_fraction

if use_watershed == True:
    PDF_save_file_name = folder / f"{active_file.stem}_TEM_Particle_Count_Report.pdf"
    xlsx_save_file_name = folder / f"{active_file.stem}_TEM_Particle_Count_Report.xlsx"
    count_type = 'particle'
else:
    PDF_save_file_name = folder / f"{active_file.stem}_TEM_Aggregate_Count_Report.pdf"
    count_type = 'aggregate'
    xlsx_save_file_name = folder / f"{active_file.stem}_TEM_Aggregate_Count_Report.xlsx"
pdf_path = PDF_save_file_name
xlsx_path = xlsx_save_file_name
steps = ["Loading & Filtering", "Segmenting Particles", "Calculating Metrics", "Generating Report"]

# Setup persistent display engine
status_display = display(display_id=True)

def print_status(step_idx):
    pct = int((step_idx / len(steps)) * 100)
    bar = "█" * (step_idx * 5) + "-" * ((len(steps) - step_idx) * 5)
    status_display.update(f"Processing Image: [{bar}] {pct}% | Current Step: {steps[step_idx-1]}")

def get_true_min_feret(region_mask):
    coords = np.argwhere(region_mask)
    if len(coords) < 3:
        return 0.0
    hull = ConvexHull(coords)
    hull_points = coords[hull.vertices]
    min_width = float('inf')
    for angle in np.linspace(0, np.pi, 90):
        rot = np.array([[np.cos(angle), -np.sin(angle)],
                        [np.sin(angle), np.cos(angle)]])
        rotated = hull_points @ rot.T
        width = rotated[:, 0].max() - rotated[:, 0].min()
        if width < min_width:
            min_width = width
    return min_width
print(f"{active_file.name}")
print(f"Loading took: {time.time() - start:.2f} s")
total_time_for_image = total_time_for_image + (time.time()-start)
start = time.time()
with PdfPages(pdf_path) as pdf:
    
    # --- STEP 1: Image Prep & Local Thresholding ---
    print_status(1)

    scale_factor = 0.1  # 10% of original size
    small_image = transform.rescale(image, scale_factor, anti_aliasing=False, preserve_range=True)
    
    # 2. Blur the tiny image (scale down the sigma proportionally!)
    small_sigma = background_blur_sigma * scale_factor
    small_bg = filters.gaussian(small_image, sigma=small_sigma)
    
    # 3. Resize back to original dimensions
    bg_map = transform.resize(small_bg, image.shape, preserve_range=True)
    flattened_image = image - bg_map

    print(f"Image Prep & Local Thresholding took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    
    # Apply Gaussian smoothing to reduce background grain noise before thresholding
    smoothed_image = filters.gaussian(flattened_image, sigma=gaussian_sigma)

    print(f"Gaussian smoothing took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    
    # Adaptive threshold adapts to local brightness changes across the carbon grid
    local_thresh = filters.threshold_local(smoothed_image, block_size=adaptive_block_size, offset=adaptive_offset, method="mean")
    binary_mask = smoothed_image < local_thresh

    print(f"Adaptive threshold took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    
    # Baseline cleaning for visualization
    binary_mask = morphology.remove_small_objects(binary_mask, min_size=50)
    fig1, axes = plt.subplots(1, 2, figsize=(15, 5), layout="constrained")
    axes[0].imshow(image, cmap='gray')
    axes[0].set_title('Original')
    axes[1].imshow(binary_mask, cmap='gray')
    axes[1].set_title('Thresholded (Adaptive Local)')
    for ax in axes:
        ax.axis('off')
    pdf.savefig(fig1)
    plt.close(fig1)
    
    print(f"Removing tiny artifacts and constructing PDF pg1 took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    
    # --- STEP 2: Morphology Operations ---
    print_status(2)
    binary_filled = ndi.binary_fill_holes(binary_mask)
    binary_clean_first = morphology.remove_small_objects(binary_filled, min_size=first_pass * min_particle_area)
    binary_glued = morphology.binary_closing(binary_clean_first, morphology.disk(aggregate_bridging_factor))
    binary_clean_final = morphology.remove_small_objects(binary_glued, min_size=min_particle_area)
    
    # Erase scale bar zone and dynamically exclude particles touching the false border
    cutoff_row = int(img_height * (1.0 - scale_bar_zone_fraction))
    
    if exclude_border_particles:
        # Run clear_border on only the top zone so the cutoff line acts as the bottom edge
        binary_clean_final[:cutoff_row, :] = clear_border(binary_clean_final[:cutoff_row, :])
        
    # Always ensure the bottom scale bar zone is completely blacked out
    binary_clean_final[cutoff_row:, :] = False
        
    if use_watershed:
        from skimage.feature import peak_local_max
        from skimage.segmentation import watershed
        
        # 1. Map out structural depth (distance from edges)
        distance_map = ndi.distance_transform_edt(binary_clean_final)
        
        # 2. Locate unique particle centers based on your min_distance parameter
        local_max_coords = peak_local_max(
            distance_map, 
            min_distance=watershed_min_distance, 
            labels=binary_clean_final
        )
        
        # 3. Label the detected peak markers
        peaks_mask = np.zeros(binary_clean_final.shape, dtype=bool)
        if len(local_max_coords) > 0:
            peaks_mask[tuple(local_max_coords.T)] = True
        markers = measure.label(peaks_mask)
        
        # 4. Flood the inverted landscape to split touching boundaries
        label_image = watershed(-distance_map, markers, mask=binary_clean_final)
    else:
        # Fallback to your original, unified aggregate measurement behavior
        label_image = measure.label(binary_clean_final)
    
    props = measure.regionprops(label_image)

    print(f"Final image morphology operations and particle detection took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    # --- STEP 3: Region Properties Loop ---
    print_status(3)
    results = []
    
    # Parse filename information safely
    try:
        sample_name = active_file.name.split("_")[0]
        mag_name = active_file.name.split('_')[1].split('(')[0].strip('x')
        mag_name = mag_name.split('.')[0]
        mag_name = mag_name + 'x'
    except IndexError:
        sample_name = "Unknown"
        mag_name = "Unknown"

    for region in props:
        y_center, x_center = region.centroid
        if y_center > (img_height * (1.0 - scale_bar_zone_fraction)):
            continue
            
        region_coords = np.argwhere(region.image)
        if len(region_coords) < 3:
            continue
            
        unique_y = np.unique(region_coords[:, 0])
        unique_x = np.unique(region_coords[:, 1])
        if len(unique_y) < 2 or len(unique_x) < 2:
            continue
            
        max_f = region.feret_diameter_max
        min_f = get_true_min_feret(region.image)
        
        hull = ConvexHull(np.argwhere(region.image))
        p_convex = hull.area
        
        results.append({
            'sample': sample_name,
            'magnification': mag_name,
            'filename': active_file.name,
            'type': count_type,
            'label': region.label,
            'area': region.area,
            'convex area': region.area_convex,
            'convex perimeter': p_convex,
            'max feret': max_f,
            'min feret': min_f,
            'circularity': (4 * np.pi * region.area_convex ) / (p_convex**2),
            'max feret / min feret': max_f / (min_f if min_f > 0 else 1),
            'solidity': region.area / region.area_convex
        })
        
    # Check if any particles were detected to avoid empty dataframe crashes
    if len(results) > 0:
        df = pd.DataFrame(results)
        df_rounded = df.round(2)
    else:
        df_rounded = pd.DataFrame(columns=['label', 'area', 'max feret', 'min feret', 'solidity'])

    print(f"Taking particle measurements and creating particle measurement database took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    # --- STEP 4: Render Plots & Table to PDF ---
    print_status(4)
    
    # PAGE 2: Detection Verification
    fig2, ax = plt.subplots(figsize=(10, 10),layout="constrained")
    ax.imshow(image, cmap='gray')
    for region in props:
        if region.area < 500:
            continue
        y0, x0 = region.centroid
        if y0 > (img_height * (1.0 - scale_bar_zone_fraction)):
            continue
        lbl = region.label
        d_max = region.feret_diameter_max
        
        ax.text(x0, y0, f"ID:{lbl}", color='yellow', fontsize=9, fontweight='bold', ha='center')
        ax.text(x0, y0 + label_offset, f"{d_max:.1f}px", color='springgreen', fontsize=8, fontweight='bold', ha='center')
        
        minr, minc, maxr, maxc = region.bbox
        rect = plt.Rectangle((minc, minr), maxc - minc, maxr - minr, fill=False, edgecolor='darkviolet', linewidth=1)
        ax.add_patch(rect)
        
    ax.set_title("Detection & Measurement Verification")
    plt.axis('off')
    pdf.savefig(fig2)
    plt.close(fig2)
    
    print(f"Confirming measurements in image bounds and above min size, PDF pg2 took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    
    # PAGE 3: Verification Map (With Perimeters)
    color_labels = color.label2rgb(label_image, image=image, bg_label=0)
    fig3, ax = plt.subplots(figsize=(12, 12),layout="constrained")
    ax.imshow(color_labels)
    
    contours = find_contours(binary_clean_final, level=0.5)
    for contour in contours:
        ax.plot(contour[:, 1], contour[:, 0], color='cyan', linewidth=1.5)
        
    for region in props:
        y0, x0 = region.centroid
        if y0 > (img_height * (1.0 - scale_bar_zone_fraction)):
            continue
        if region.area < min_particle_area:
            continue
        ax.text(x0, y0, str(region.label), color='white', fontsize=10, fontweight='bold', ha='center', va='center')
        
    ax.set_title("Verification Map (With Outlined Perimeters)")
    plt.axis('off')
    pdf.savefig(fig3)
    plt.close(fig3)

status_display.update(f"Optimization Complete! Single-image report compiled: {pdf_path.name}")
if 'df' in locals() or 'df' in globals():
    df.to_excel(xlsx_path, index=False)

print(f"Particle measurements verification map, PDF pg3 took: {time.time() - start:.2f} s")
total_time_for_image = total_time_for_image + (time.time()-start)

total_time_minute_fraction = total_time_for_image / 60
total_time_minute = int(total_time_minute_fraction)
total_time_seconds = (total_time_minute_fraction - total_time_minute) * 60
print(f"Total analysis time: {total_time_minute} min, {round(total_time_seconds,2)} s")
```

```python
current_file = 1
total_files = len(list(folder.glob("*.jp*g")))

for active_file in folder.glob("*.jp*g"):
    print(f"\n(File {current_file} of {total_files})")
    current_file += 1
    
    # --- 3. GEOMETRY SUGGESTION ENGINE ---
    _img_raw = io.imread(active_file)
    _img_temp = color.rgb2gray(_img_raw) if _img_raw.ndim == 3 else _img_raw
    _img_temp = util.img_as_float32(_img_temp)
    img_height, img_width = _img_temp.shape
    
    _test_mask = _img_temp < filters.threshold_otsu(_img_temp)
    _dist_transform = ndi.distance_transform_edt(_test_mask)
    _max_thickness = np.max(_dist_transform)
    
    _suggested_bg_blur = int(max(img_width, img_height) * 0.10)
    _suggested_block = int(_max_thickness * 4)
    if _suggested_block % 2 == 0: _suggested_block += 1
    _suggested_block = max(21, min(_suggested_block, 151))
    
    _suggested_offset = round(float(np.std(_img_temp) * 0.12), 3)
    _est_particle_radius = max(3.0, _max_thickness * 0.5)
    _suggested_min_area = int((_est_particle_radius ** 2) * np.pi)
    _suggested_watershed_dist = int(_est_particle_radius * 1.5)
    
    print("="*60)
    print(f"   📊 SUGGESTED BASELINES FOR: {active_file.name}")
    print("="*60)
    print(f" -> background_blur_sigma: {_suggested_bg_blur} px")
    print(f" -> gaussian_sigma:         1")
    print(f" -> adaptive_block_size:    {_suggested_block} px")
    print(f" -> adaptive_offset:        {_suggested_offset}")
    print(f" -> min_particle_area:      {_suggested_min_area} px")
    print(f" -> watershed_min_distance: {_suggested_watershed_dist} px")
    print("="*60 + "\n")
    
    # =======================# 2. TUNABLE PARAMETERS====================================================
    adaptive_offset =  _suggested_offset * 1.3                              # if offset = 0 a particle just must be above local mean (Threshold = local mean - offset)
    min_particle_area = int(_suggested_min_area * 0.6)
    first_pass = .4
    aggregate_bridging_factor = 10                                           # higher numbers bridge more gaps
    
    adaptive_block_size = int(_suggested_block * 1)                         # pixel diameter for relative brightness [otsu would be adaptive_block_size = whole image and adaptive_offset = 0]
    if adaptive_block_size % 2 == 0: adaptive_block_size += 1
    
    # Image smoothing
    gaussian_sigma = 1                                                      # Higher values smooth rough surfaces and blend noise
    background_blur_sigma = _suggested_bg_blur                              # pixel diameer, Must > largest aggregate size, subtracts local background
      
    # watershed particle separation [only toggle for particles]
    use_watershed = False                                                    # True = slice clusters into particles
    watershed_min_distance = int(_suggested_watershed_dist * 1)             # particles wont be sliced up so they their centers are closer than this distance
    
    # Data Integrity Controls
    exclude_border_particles = True                                        # True removes chopped particles touching image borders
    scale_bar_zone_fraction = 0.03                                         # Fraction of the bottom screen to erase (0.05 = bottom 5%)
    
    # ==================================================================================================
    # 3. IMAGE PROCESSING & MEASUREMENT PIPELINE
    # ==================================================================================================
    start = time.time()
    total_time_for_image = 0
    
    image_raw = io.imread(active_file)
    if image_raw.ndim == 3:
        image = color.rgb2gray(image_raw)
    else:
        image = image_raw
    image = util.img_as_float32(image)
    img_height, img_width = image.shape
    label_offset = img_height * scale_bar_zone_fraction
    
    if use_watershed == True:
        PDF_save_file_name = folder / f"{active_file.stem}_TEM_Particle_Count_Report.pdf"
        xlsx_save_file_name = folder / f"{active_file.stem}_TEM_Particle_Count_Report.xlsx"
        count_type = 'particle'
    else:
        PDF_save_file_name = folder / f"{active_file.stem}_TEM_Aggregate_Count_Report.pdf"
        count_type = 'aggregate'
        xlsx_save_file_name = folder / f"{active_file.stem}_TEM_Aggregate_Count_Report.xlsx"
    pdf_path = PDF_save_file_name
    xlsx_path = xlsx_save_file_name
    steps = ["Loading & Filtering", "Segmenting Particles", "Calculating Metrics", "Generating Report"]
    
    # Setup persistent display engine
    status_display = display(display_id=True)
    
    def print_status(step_idx):
        pct = int((step_idx / len(steps)) * 100)
        bar = "█" * (step_idx * 5) + "-" * ((len(steps) - step_idx) * 5)
        status_display.update(f"Processing Image: [{bar}] {pct}% | Current Step: {steps[step_idx-1]}")
    
    def get_true_min_feret(region_mask):
        coords = np.argwhere(region_mask)
        if len(coords) < 3:
            return 0.0
        hull = ConvexHull(coords)
        hull_points = coords[hull.vertices]
        min_width = float('inf')
        for angle in np.linspace(0, np.pi, 90):
            rot = np.array([[np.cos(angle), -np.sin(angle)],
                            [np.sin(angle), np.cos(angle)]])
            rotated = hull_points @ rot.T
            width = rotated[:, 0].max() - rotated[:, 0].min()
            if width < min_width:
                min_width = width
        return min_width
    print(f"{active_file.name}")
    print(f"Loading took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    start = time.time()
    with PdfPages(pdf_path) as pdf:
        
        # --- STEP 1: Image Prep & Local Thresholding ---
        print_status(1)
    
        scale_factor = 0.1  # 10% of original size
        small_image = transform.rescale(image, scale_factor, anti_aliasing=False, preserve_range=True)
        
        # 2. Blur the tiny image (scale down the sigma proportionally!)
        small_sigma = background_blur_sigma * scale_factor
        small_bg = filters.gaussian(small_image, sigma=small_sigma)
        
        # 3. Resize back to original dimensions
        bg_map = transform.resize(small_bg, image.shape, preserve_range=True)
        flattened_image = image - bg_map
    
        print(f"Image Prep & Local Thresholding took: {time.time() - start:.2f} s")
        total_time_for_image = total_time_for_image + (time.time()-start)
        start = time.time()
        
        # Apply Gaussian smoothing to reduce background grain noise before thresholding
        smoothed_image = filters.gaussian(flattened_image, sigma=gaussian_sigma)
    
        print(f"Gaussian smoothing took: {time.time() - start:.2f} s")
        total_time_for_image = total_time_for_image + (time.time()-start)
        start = time.time()
        
        # Adaptive threshold adapts to local brightness changes across the carbon grid
        local_thresh = filters.threshold_local(smoothed_image, block_size=adaptive_block_size, offset=adaptive_offset, method="mean")
        binary_mask = smoothed_image < local_thresh
    
        print(f"Adaptive threshold took: {time.time() - start:.2f} s")
        total_time_for_image = total_time_for_image + (time.time()-start)
        start = time.time()
        
        # Baseline cleaning for visualization
        binary_mask = morphology.remove_small_objects(binary_mask, min_size=50)
        fig1, axes = plt.subplots(1, 2, figsize=(15, 5), layout="constrained")
        axes[0].imshow(image, cmap='gray')
        axes[0].set_title('Original')
        axes[1].imshow(binary_mask, cmap='gray')
        axes[1].set_title('Thresholded (Adaptive Local)')
        for ax in axes:
            ax.axis('off')
        pdf.savefig(fig1)
        plt.close(fig1)
        
        print(f"Removing tiny artifacts and constructing PDF pg1 took: {time.time() - start:.2f} s")
        total_time_for_image = total_time_for_image + (time.time()-start)
        start = time.time()
        
        # --- STEP 2: Morphology Operations ---
        print_status(2)
        binary_filled = ndi.binary_fill_holes(binary_mask)
        binary_clean_first = morphology.remove_small_objects(binary_filled, min_size=first_pass * min_particle_area)
        binary_glued = morphology.binary_closing(binary_clean_first, morphology.disk(aggregate_bridging_factor))
        binary_clean_final = morphology.remove_small_objects(binary_glued, min_size=min_particle_area)
        
        # Erase scale bar zone and dynamically exclude particles touching the false border
        cutoff_row = int(img_height * (1.0 - scale_bar_zone_fraction))
        
        if exclude_border_particles:
            # Run clear_border on only the top zone so the cutoff line acts as the bottom edge
            binary_clean_final[:cutoff_row, :] = clear_border(binary_clean_final[:cutoff_row, :])
            
        # Always ensure the bottom scale bar zone is completely blacked out
        binary_clean_final[cutoff_row:, :] = False
            
        if use_watershed:
            from skimage.feature import peak_local_max
            from skimage.segmentation import watershed
            
            # 1. Map out structural depth (distance from edges)
            distance_map = ndi.distance_transform_edt(binary_clean_final)
            
            # 2. Locate unique particle centers based on your min_distance parameter
            local_max_coords = peak_local_max(
                distance_map, 
                min_distance=watershed_min_distance, 
                labels=binary_clean_final
            )
            
            # 3. Label the detected peak markers
            peaks_mask = np.zeros(binary_clean_final.shape, dtype=bool)
            if len(local_max_coords) > 0:
                peaks_mask[tuple(local_max_coords.T)] = True
            markers = measure.label(peaks_mask)
            
            # 4. Flood the inverted landscape to split touching boundaries
            label_image = watershed(-distance_map, markers, mask=binary_clean_final)
        else:
            # Fallback to your original, unified aggregate measurement behavior
            label_image = measure.label(binary_clean_final)
        
        props = measure.regionprops(label_image)
    
        print(f"Final image morphology operations and particle detection took: {time.time() - start:.2f} s")
        total_time_for_image = total_time_for_image + (time.time()-start)
        start = time.time()
        # --- STEP 3: Region Properties Loop ---
        print_status(3)
        results = []
        
        # Parse filename information safely
        try:
            sample_name = active_file.name.split("_")[0]
            mag_name = active_file.name.split('_')[1].split('(')[0].strip('x')
            mag_name = mag_name.split('.')[0]
            mag_name = mag_name + 'x'
        except IndexError:
            sample_name = "Unknown"
            mag_name = "Unknown"
    
        for region in props:
            y_center, x_center = region.centroid
            if y_center > (img_height * (1.0 - scale_bar_zone_fraction)):
                continue
                
            region_coords = np.argwhere(region.image)
            if len(region_coords) < 3:
                continue
                
            unique_y = np.unique(region_coords[:, 0])
            unique_x = np.unique(region_coords[:, 1])
            if len(unique_y) < 2 or len(unique_x) < 2:
                continue
                
            max_f = region.feret_diameter_max
            min_f = get_true_min_feret(region.image)
            
            hull = ConvexHull(np.argwhere(region.image))
            p_convex = hull.area
            
            results.append({
                'sample': sample_name,
                'magnification': mag_name,
                'filename': active_file.name,
                'type': count_type,
                'label': region.label,
                'area': region.area,
                'convex area': region.area_convex,
                'convex perimeter': p_convex,
                'max feret': max_f,
                'min feret': min_f,
                'circularity': (4 * np.pi * region.area_convex ) / (p_convex**2),
                'max feret / min feret': max_f / (min_f if min_f > 0 else 1),
                'solidity': region.area / region.area_convex
            })
            
        # Check if any particles were detected to avoid empty dataframe crashes
        if len(results) > 0:
            df = pd.DataFrame(results)
            df_rounded = df.round(2)
        else:
            df_rounded = pd.DataFrame(columns=['label', 'area', 'max feret', 'min feret', 'solidity'])
    
        print(f"Taking particle measurements and creating particle measurement database took: {time.time() - start:.2f} s")
        total_time_for_image = total_time_for_image + (time.time()-start)
        start = time.time()
        # --- STEP 4: Render Plots & Table to PDF ---
        print_status(4)
        
        # PAGE 2: Detection Verification
        fig2, ax = plt.subplots(figsize=(10, 10),layout="constrained")
        ax.imshow(image, cmap='gray')
        for region in props:
            if region.area < 500:
                continue
            y0, x0 = region.centroid
            if y0 > (img_height * (1.0 - scale_bar_zone_fraction)):
                continue
            lbl = region.label
            d_max = region.feret_diameter_max
            
            ax.text(x0, y0, f"ID:{lbl}", color='yellow', fontsize=9, fontweight='bold', ha='center')
            ax.text(x0, y0 + label_offset, f"{d_max:.1f}px", color='springgreen', fontsize=8, fontweight='bold', ha='center')
            
            minr, minc, maxr, maxc = region.bbox
            rect = plt.Rectangle((minc, minr), maxc - minc, maxr - minr, fill=False, edgecolor='darkviolet', linewidth=1)
            ax.add_patch(rect)
            
        ax.set_title("Detection & Measurement Verification")
        plt.axis('off')
        pdf.savefig(fig2)
        plt.close(fig2)
        
        print(f"Confirming measurements in image bounds and above min size, PDF pg2 took: {time.time() - start:.2f} s")
        total_time_for_image = total_time_for_image + (time.time()-start)
        start = time.time()
        
        # PAGE 3: Verification Map (With Perimeters)
        color_labels = color.label2rgb(label_image, image=image, bg_label=0)
        fig3, ax = plt.subplots(figsize=(12, 12),layout="constrained")
        ax.imshow(color_labels)
        
        contours = find_contours(binary_clean_final, level=0.5)
        for contour in contours:
            ax.plot(contour[:, 1], contour[:, 0], color='cyan', linewidth=1.5)
            
        for region in props:
            y0, x0 = region.centroid
            if y0 > (img_height * (1.0 - scale_bar_zone_fraction)):
                continue
            if region.area < min_particle_area:
                continue
            ax.text(x0, y0, str(region.label), color='white', fontsize=10, fontweight='bold', ha='center', va='center')
            
        ax.set_title("Verification Map (With Outlined Perimeters)")
        plt.axis('off')
        pdf.savefig(fig3)
        plt.close(fig3)
    
    status_display.update(f"Optimization Complete! Single-image report compiled: {pdf_path.name}")
    if 'df' in locals() or 'df' in globals():
        df.to_excel(xlsx_path, index=False)
    
    print(f"Particle measurements verification map, PDF pg3 took: {time.time() - start:.2f} s")
    total_time_for_image = total_time_for_image + (time.time()-start)
    
    total_time_minute_fraction = total_time_for_image / 60
    total_time_minute = int(total_time_minute_fraction)
    total_time_seconds = (total_time_minute_fraction - total_time_minute) * 60
    print(f"Total analysis time: {total_time_minute} min, {round(total_time_seconds,2)} s")
print("---------------------------------------------   Finished with all files in folder   ---------------------------------------------")
```
